```python
from tkinter import *
from tkinter import ttk, messagebox
from gtts import gTTS
from playsound3 import playsound
import os
import threading
```

```python
# Modern window setup
root = Tk()
root.title("Tamil Text-to-Speech Studio")
root.geometry("650x470")
root.minsize(600, 430)
root.configure(bg="#F3F6FB")
root.resizable(True, True)

# Colors
BG = "#F3F6FB"
CARD = "#FFFFFF"
DARK = "#172554"
MUTED = "#64748B"
GREEN = "#0F9D78"
RED = "#E5484D"
BLUE = "#2563EB"
```

```python
# Header
header = Frame(root, bg=BG)
header.pack(fill="x", padx=28, pady=(24, 14))

Label(header, text="🎙  Tamil Voice Studio", font=("Segoe UI", 22, "bold"),
      bg=BG, fg=DARK).pack(anchor="w")
Label(header, text="Turn your Tamil text into spoken audio", font=("Segoe UI", 11),
      bg=BG, fg=MUTED).pack(anchor="w", pady=(5, 0))

# Main card
card = Frame(root, bg=CARD, highlightbackground="#E2E8F0", highlightthickness=1)
card.pack(fill="both", expand=True, padx=28, pady=(0, 16))

Label(card, text="Your narration", font=("Segoe UI", 13, "bold"),
      bg=CARD, fg=DARK).pack(anchor="w", padx=22, pady=(20, 8))
Label(card, text="Type or paste Tamil text below", font=("Segoe UI", 10),
      bg=CARD, fg=MUTED).pack(anchor="w", padx=22, pady=(0, 10))

Msg = StringVar()
entry_field = Text(card, height=6, wrap="word", font=("Nirmala UI", 13),
                   bg="#F8FAFC", fg="#0F172A", relief="flat",
                   padx=12, pady=10, insertbackground=DARK,
                   highlightthickness=1, highlightbackground="#CBD5E1",
                   highlightcolor=BLUE)
entry_field.pack(fill="both", expand=True, padx=22, pady=(0, 16))

# Action buttons
actions = Frame(card, bg=CARD)
actions.pack(fill="x", padx=22, pady=(0, 18))

status_var = StringVar(value="Ready • Enter text to begin")
status_label = Label(root, textvariable=status_var, font=("Segoe UI", 10),
                     bg=BG, fg=MUTED, anchor="w")
status_label.pack(fill="x", padx=30, pady=(0, 18))
```

```python
def Text_to_speech():
    message = entry_field.get("1.0", "end-1c").strip()
    if not message:
        messagebox.showwarning("Text required", "Please enter some Tamil text first.")
        entry_field.focus_set()
        return

    def generate_audio():
        try:
            status_var.set("Generating audio… Please wait")
            root.after(0, lambda: status_label.config(fg=BLUE))
            speech = gTTS(text=message, lang="ta")
            speech.save("DataFlair.mp3")
            root.after(0, lambda: status_var.set("Playing your Tamil audio…"))
            playsound("DataFlair.mp3")
            root.after(0, lambda: status_var.set("Finished • Audio saved as DataFlair.mp3"))
            root.after(0, lambda: status_label.config(fg=GREEN))
        except Exception as error:
            root.after(0, lambda: status_var.set("Something went wrong. Check your internet connection."))
            root.after(0, lambda: status_label.config(fg=RED))
            root.after(0, lambda: messagebox.showerror("Text-to-Speech error", str(error)))

    threading.Thread(target=generate_audio, daemon=True).start()
```

```python
def Exit():
    root.destroy()
```

```python
def Reset():
    entry_field.delete("1.0", "end")
    status_var.set("Ready • Enter text to begin")
    status_label.config(fg=MUTED)
    entry_field.focus_set()
```

```python
# PLAY button
Button(actions, text="▶  PLAY AUDIO", font=("Segoe UI", 11, "bold"),
       fg="white", bg=GREEN, activebackground="#0B8061", activeforeground="white",
       relief="flat", bd=0, cursor="hand2", padx=18, pady=11,
       command=Text_to_speech).pack(side="left")

# RESET button
Button(actions, text="↻  RESET", font=("Segoe UI", 11, "bold"),
       fg=BLUE, bg="#EAF1FF", activebackground="#D7E5FF", activeforeground=BLUE,
       relief="flat", bd=0, cursor="hand2", padx=18, pady=11,
       command=Reset).pack(side="left", padx=(10, 0))

# EXIT button
Button(actions, text="✕  EXIT", font=("Segoe UI", 11, "bold"),
       fg=RED, bg="#FFF0F0", activebackground="#FFE0E0", activeforeground=RED,
       relief="flat", bd=0, cursor="hand2", padx=18, pady=11,
       command=Exit).pack(side="right")

entry_field.focus_set()
root.mainloop()
```
